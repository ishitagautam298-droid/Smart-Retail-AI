```python
import json

intent_path = "/content/drive/MyDrive/Smart-Retail-AI/datasets/intents.json"

with open(intent_path, "r") as file:
    intents_data = json.load(file)

print("Intents loaded successfully!")
```

```python
patterns = []
tags = []

for intent in intents_data["intents"]:
    for pattern in intent["patterns"]:
        patterns.append(pattern)
        tags.append(intent["tag"])

print("Total Questions:", len(patterns))
print("Total Intents:", len(set(tags)))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

chatbot_vectorizer = TfidfVectorizer()

X = chatbot_vectorizer.fit_transform(patterns)

print("Feature Shape:", X.shape)
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y = label_encoder.fit_transform(tags)

print(y)
```

```python
from sklearn.linear_model import LogisticRegression

chatbot_model = LogisticRegression(
    max_iter=1000
)

chatbot_model.fit(
    X,
    y
)

print("Chatbot model trained successfully!")
```

```python
import random

def chatbot_response(message):

    # Convert user message into vector
    message_vector = chatbot_vectorizer.transform([message])

    # Predict intent
    prediction = chatbot_model.predict(message_vector)

    # Convert number back to intent name
    intent_tag = label_encoder.inverse_transform(prediction)[0]


    # Find matching response
    for intent in intents_data["intents"]:

        if intent["tag"] == intent_tag:
            return random.choice(intent["responses"])


    return "Sorry, I could not understand your query."
```
